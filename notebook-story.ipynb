# The Night the Pallets Went Missing
*A short story told in markdown and Python*

At 6 a.m., Dana, a logistics analyst at a small beverage distributor, opens her laptop and finds a message: **"Warehouse says 240 pallets are missing from the Friday shipment."**

She has one tool and one hour: a Jupyter notebook. This is the notebook she wrote.

## Part 1 - The Setup

Dana's company ships pallets of drinks from one warehouse to several stores. She starts by writing down what she knows.

![Python logo](https://www.python.org/static/community_logos/python-logo.png)

She works in Python, using [Jupyter](https://jupyter.org) so she can mix notes and calculations in one place.

```python
# What the shipping paperwork says
warehouse = "Central Warehouse"
pallets_loaded = 1200
cases_per_pallet = 48
price_per_case = 36.50

print(f"{warehouse} loaded {pallets_loaded} pallets on Friday.")
```

## Part 2 - The Problem

Each store confirmed what it received. Dana types the numbers in and compares them to what was loaded.

```python
received = {
    "Downtown": 310,
    "Harbor": 275,
    "Airport": 190,
    "Hillside": 185,
}

total_received = sum(received.values())
missing = pallets_loaded - total_received

print("Total received:", total_received)
print("Missing pallets:", missing)
```

```python
# How much money is that?
lost_cases = missing * cases_per_pallet
lost_value = lost_cases * price_per_case

print(f"Missing cases: {lost_cases}")
print(f"Value at risk: ${lost_value:,.2f}")
```

That is a big number. Before anyone panics, Dana wants to know **where** the shortfall is, not only how large it is.

## Part 3 - The Investigation

Each store was supposed to receive a planned amount. She compares plan against reality, store by store.

```python
planned = {
    "Downtown": 310,
    "Harbor": 275,
    "Airport": 190,
    "Hillside": 425,
}

for store in planned:
    gap = planned[store] - received[store]
    status = "OK" if gap == 0 else f"SHORT by {gap}"
    print(f"{store:10} planned={planned[store]:4}  received={received[store]:4}  -> {status}")
```

```python
# Which store is the culprit?
gaps = {store: planned[store] - received[store] for store in planned}
culprit = max(gaps, key=gaps.get)

print(f"All of the shortfall is at: {culprit} ({gaps[culprit]} pallets)")
```

Three stores are perfect. **Hillside** is short by exactly 240 pallets, which is one full truck. The pallets were never lost: the Hillside truck probably never left the warehouse.

## Part 4 - The Resolution

Dana calls the dock. The truck was parked behind a loading door and nobody noticed. It leaves now, and she calculates the delay.

```python
truck_capacity = 240
hours_late = 5
hours_of_shelf_life_used = hours_late / 24

print(f"Truck carries {truck_capacity} pallets, {hours_late} hours late.")
print(f"Shelf life used up by the delay: {hours_of_shelf_life_used:.1%} of one day")
```

```python
# Final report for the manager
report = f'''
Status: RESOLVED
Missing at 6 a.m.: {missing} pallets (${lost_value:,.2f})
Cause: Hillside truck never departed
Delay: {hours_late} hours, no product lost
'''
print(report)
```

## The End

By 7 a.m. the truck was on the road and the manager had a one-paragraph report. Nothing was lost, and the mystery took about twenty lines of Python.

**What Dana learned:** a notebook is not only for running code. It keeps the question, the evidence and the answer in one readable story.
